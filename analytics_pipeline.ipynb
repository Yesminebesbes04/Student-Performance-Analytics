```python
import pandas as pd
df= pd.read_csv("student_performance_messy.csv")
```

```python
def clean_data(df):
  df=df.drop_duplicates()
  cols=["study_hrs","attend","prev"]
  for c in cols:
    df[c]=df[c].str.replace(r"hrs|%|/100","")
    df[c]=pd.to_numeric(df[c])
  df=df[(df.attend<=100)&(df.exam<=100)]
  df=df.fillna(df.median(numeric_only=True))
  return df
```

```python
def transform_data(df):

  df["grade"]=pd.cut(df.exam,
      bins=[0,49,79,100],
      labels=["NeedsImprovement","Good","Excellent"])

  df["att_group"]=pd.cut(df.attend,
      bins=[0,75,90,100],
      labels=["Low","Mid","High"])
  return df
```

```python
def analyze_data(df):

  summary=df.groupby("att_group").agg(
      avg_study_hrs=("study_hrs","mean"),
      avg_score=("exam","mean"),
      students=("id","count"))
  corr=df.corr(numeric_only=True)
  return summary, corr
```

```python
def load_results(df,summary):
  df.to_csv("clean_students.cvs")
  summary.to_csv("analysis_results.cvs")
```

```python
def run_pipeline():
  df=pd.read_csv("student_performance_messy.csv")
  df=clean_data(df)
  df=transform_data(df)
  summary,corr=analyze_data(df)
  load_results(df,summary)
```
